```python
import tensorflow as tf #builds the model
from tensorflow.keras.models import Sequential #execute command one after another
from tensorflow.keras.layers import Dense,Flatten
from tensorflow.keras.datasets import mnist
```

```python
(x_train,y_train,), (x_test,y_test)=mnist.load_data()
```

```python
#normalize the input
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
# Build the model
model=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
# compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train the model
model.fit(x_train,y_train,epochs=6)
```

```python
#Evaluate the model
loss, acc = model.evaluate(x_test, y_test)
print("Accuracy:", acc)
```
